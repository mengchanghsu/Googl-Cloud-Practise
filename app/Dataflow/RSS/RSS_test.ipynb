```python
"""
RSS訂閱
"""
import feedparser
import requests
import ssl
import pandas as pd
from datetime import datetime

rss_url = 'http://feeds.feedburner.com/cnaFirstNews'
```

```python
feed = feedparser.parse(rss_url)

# 文章來源標題
print(f"Feed Title: {feed.feed.title}")

# 資料處理
df = pd.DataFrame(feed.entries)
df.iloc[0][['id', 'guidislink', 'title', 'link', 'published', 'published_parsed', 'summary']]
```

```python
# published_parsed: 年、月、日、时、分、秒、一周的第几天（0 是周一，6 是周日）、一年中的第几天（从 1 到 365）、是否为夏令时
published = df.iloc[0]['published_parsed']
published_date = datetime(published.tm_year, published.tm_mon, published.tm_mday, published.tm_hour, published.tm_min, published.tm_sec)
print(f"Published Date: {published_date}")
```

```python
rss_url = 'https://news.cnyes.com/rss/v1/news/category/headline'

ssl._create_default_https_context = ssl._create_unverified_context  # 忽略SSL證書檢查
feed = feedparser.parse(rss_url)

# 資料處理
df = pd.DataFrame(feed.entries)
df.iloc[0][['id', 'guidislink', 'title', 'link', 'published', 'published_parsed', 'summary', 'tags', 'media_keywords']]
```

```python
df_tmp = df[df['media_keywords'].str.contains('盤中速報', na=False) == False]
df_tmp = df_tmp[df_tmp['media_keywords'].str.contains('市場預估', na=False) == False]
df_tmp = df_tmp[df_tmp['media_keywords'].str.contains('台股盤後', na=False) == False]

str_context = ""
for r in df_tmp['summary']:
    str_context += r + '\n'

print(str_context)
```
