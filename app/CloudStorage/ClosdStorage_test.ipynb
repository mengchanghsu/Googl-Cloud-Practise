```python
"""
cloud storage module 操作練習

# 安裝library
pip install --upgrade google-cloud-storage

# 登入才能執行python，不啟動瀏覽器加 --no-browser
gcloud auth application-default login
"""
from google.cloud import storage

# 初始化
storage_client = storage.Client()
```

```python
import os
os.system("gcloud auth application-default print-access-token")
```

```python
# 上傳, 儲存桶路徑 your-bucket/folder1/file.txt
bucket_name = 'your-bucket'
file_path = 'local端檔案路徑'
bucket = storage_client.bucket(bucket_name)
blob = bucket.blob('folder1/file.txt')  # 由於平面命名空間,把子目錄路徑寫在一起做為物件名稱
blob.upload_from_filename(file_path)    # 上傳
```
